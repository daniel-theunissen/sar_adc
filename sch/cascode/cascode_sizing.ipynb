# Regulated-cascode devices

First-cut widths of the NMOS (T1, T3) and PMOS (T5, T7) cascode devices.

```python
from sarlib import specs as S
from sarlib.gmid import nmos, pmos, size, device_table

I_FULL = 1e6 * S.I_BIAS                    # uA, current source copies
I_MID = 1e6 * (S.I_BIAS - S.I_FS / 2)      # uA, cascode device at mid code
# (ID uA, gm/Id, L um, VDS V, VSB V)
devices = dict(T3=(nmos, I_FULL, 12, 2, 0.25, 0.0), T1=(nmos, I_MID, 12, 2, 0.5, 0.25),
               T7=(pmos, I_FULL, 8, 2, 0.25, 0.0), T5=(pmos, I_MID, 12, 1, 0.2, 0.25))
device_table([(name, size(*args)) for name, args in devices.items()])
```
