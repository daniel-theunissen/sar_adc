# PMOS-input 5T OTA: channel-length sweep

```python
import numpy as np
import pandas as pd
from sarlib import specs as S
from sarlib.gmid import nmos, pmos, size

ID = 5.0                                         # uA per side
GMID = dict(pair=14, tail=8, load=8)
VDS = dict(pair=0.65, tail=0.35, load=0.8)
```

```python
rows = []
for L in (0.5, 1, 2, 4):
    p = size(pmos, ID, GMID['pair'], L, VDS['pair'])
    t = size(pmos, 2 * ID, GMID['tail'], L, VDS['tail'])
    n = size(nmos, ID, GMID['load'], L, VDS['load'])
    rows.append(dict(L=L, W_pair=p['W'], W_tail=t['W'], W_load=n['W'], gm_uS=p['gm'], gds_uS=p['gds'],
                     gain_dB=20 * np.log10(p['gm'] / (p['gds'] + n['gds'])),
                     out_lo=n['VDSAT'], out_hi=S.VDD - VDS['tail'] - p['VDSAT']))
pd.DataFrame(rows).round(3)
```
