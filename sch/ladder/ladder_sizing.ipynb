# Ladder unit device

On-resistance of the NMOS ladder switch over its source-voltage range.

```python
import numpy as np
import pandas as pd
from sarlib import specs as S
from sarlib.gmid import nmos

lad = S.ladder
```

```python
rows = []
for vs in np.linspace(*lad.VS, lad.N_VS):
    kw = dict(VGS=lad.VGATE - vs, VDS=lad.VDS, VSB=vs, L=lad.L)
    r = lad.VDS / (float(nmos.look_up('ID_W', **kw)) * 1e-6) / lad.W
    vov = lad.VGATE - vs - float(nmos.look_up('VT', **kw))
    rows.append(dict(VS=vs, R_ohm=r, Vov=vov))
df = pd.DataFrame(rows)
print(f'mean R = {df.R_ohm.mean():.0f} ohm at W = {lad.W:g} um, L = {lad.L:g} um')
df.round(3)
```
