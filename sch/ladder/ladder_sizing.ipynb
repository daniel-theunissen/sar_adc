# Ladder unit device

On-resistance of the NMOS ladder switch over its source-voltage range.

```python
import numpy as np
import pandas as pd
from sarlib.gmid import nmos

VGATE, VDS = 1.8, 0.05
L, W = 4, 1.0
VS = np.linspace(0.25, 0.85, 8)       # source-voltage range of the unit devices
```

```python
rows = []
for vs in VS:
    kw = dict(VGS=VGATE - vs, VDS=VDS, VSB=vs, L=L)
    r = VDS / (float(nmos.look_up('ID_W', **kw)) * 1e-6) / W
    vov = VGATE - vs - float(nmos.look_up('VT', **kw))
    rows.append(dict(VS=vs, R_ohm=r, Vov=vov))
df = pd.DataFrame(rows)
print(f'mean R = {df.R_ohm.mean():.0f} ohm at W = {W:g} um, L = {L:g} um')
df.round(3)
```
